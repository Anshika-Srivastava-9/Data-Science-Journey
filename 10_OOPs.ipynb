```python
class Employee:
    company="HP"
    def __init__(self,name,salary):
        self.name=name
        self.salary=salary
    def printDetails(self):
        print(f"The company of {self.name} is {self.company} with salary ${self.salary}")


    @staticmethod
    def printTime():
        print("The time is ...")


    @classmethod
    def printClassDetail(cls):
        print(f"The company is {cls.company}")
        
e=Employee("Anshika",2000000)
e.printDetails()
e.printTime()
e.printClassDetail()
```
