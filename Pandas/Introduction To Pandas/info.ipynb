```python
# selecting multiple columns 
import pandas as pd
data = {
    "name":['Abhay','Shiva','Akanksha','Anshika'],
    "age":[20,19,25,23],
    "Salary":[700000000,5500000000,900000000,800000000],
    "Peformance Score":[100,100,100,100]
}
df = pd.DataFrame(data)
print(df[["name","age"]])
```

```python
  # rows filtering
import pandas as pd
data = {
    "name":['Abhay','Shiva','Akanksha','Anshika'],
    "age":[20,19,25,23],
    "Salary":[700000000,5500000000,900000000,800000000],
    "Peformance Score":[100,100,100,100]
}
df = pd.DataFrame(data)
high_salary = df[(df["Salary"]>700000000) & (df["age"]>19)]
print(high_salary)
high_salary = df[(df["Salary"]>700000000) | (df["age"]>19)]
print(high_salary)
```
