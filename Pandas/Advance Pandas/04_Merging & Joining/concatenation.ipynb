```python
import pandas as pd 
df_customers1 = pd.DataFrame({
    'Customer_Id': [1,2,3],
    'Name': ['Silpa','Bipasha','Tammana'],
})

df_customers2 = pd.DataFrame({
    'Customer_Id': [4,5,6],
    'Name': ['Faiz','Fazal','Aamna'],
})


# Vetically Concatenation
df_concat= pd.concat([df_customers1,df_customers2], axis =0, ignore_index=True)
print('vetical concatenation')
print(df_concat)
print('\n')
df_concat= pd.concat([df_customers1,df_customers2], axis =1, ignore_index=True)
print('horizontal concatenation')
print(df_concat)



```
