'''syntax -->
pd.merge(df1,df2, on ="column_name", how ="type of join")''' 

```python
import pandas as pd 

# Customer DataFrame 
df_customers = pd.DataFrame({
    'Customer_Id': [1,2,3],
    'Name': ['Silpa','Bipasha','Tammana'],
})


# Order DataFrame
df_orders = pd.DataFrame({
    'Customer_Id': [1,2,4],
    'Order_Amount': [250, 450, 350]
})

# Merge 
df_merged = pd.merge(df_customers,df_orders, on='Customer_Id', how="inner")
print('inner join')
print(df_merged)         # return only those element which is common in both dataFrames
print("\n")

df_merged = pd.merge(df_customers,df_orders, on='Customer_Id', how="outer")
print('outer  join')
print(df_merged)        # return all element which of dataFrames and return NAN where the values are not defined 
print("\n")

df_merged = pd.merge(df_customers,df_orders, on='Customer_Id', how="right")
print('right join')
print(df_merged)       # return all element which of dataFrames from right/ second dataframe and return NAN where the values are not defined 
print("\n")

df_merged = pd.merge(df_customers,df_orders, on='Customer_Id', how="left")
print('left join')
print(df_merged)       # return all element which of dataFrames from left/ first dataframe and return NAN where the values are not defined 
print("\n")


    
```
