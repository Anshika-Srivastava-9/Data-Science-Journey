```python
import pandas as pd 
data = {
    'name':['Karan','Taran','Aman','Bhupendra','Satendra'],
    'age':[25,29,22,29,25],
    'score':[30,29,25,39,43]
}
df= pd.DataFrame(data)
grouped = df.groupby(['age'])['score'].sum()
print(grouped)
```

common aggregation methods 
1- sum() --> return sum of elements 
2- mean()--> return mean /average of data 
3- count()--> return count of not null numbers 
4- min()--> return minimum value 
5- max() --> return maximum value 
6- std() --> calculate standard deviation
