```python
# sorting using any particular column title 

import pandas as pd 
data = {
    'name':['Karan','Taran','Aman'],
    'age':[25,29,22],
    'score':[30,29,25]
}
df= pd.DataFrame(data)
print(df)

df.sort_values(by='age',ascending=True,inplace=True)
print('\n')
print('Sorted by age')
print(df)
```
