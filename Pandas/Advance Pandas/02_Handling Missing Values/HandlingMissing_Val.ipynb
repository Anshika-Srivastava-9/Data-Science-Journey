```python
# remove  the missing values
import pandas as pd 
data = {
    'name':['Arnav',None,'Rudransh','Ayansh','Pratyush','Snehal'],
    'age':[23,None,24,26,25,29],
    'salary':[20000,None,45000,60000,34000,69000],
    'performance_score':[10,None,30,60,20,40]
}
df =pd.DataFrame(data)
df.dropna(inplace=True)
print(df)
 
```

```python
# filling the missing values 
import pandas as pd 
data = {
    'name':['Arnav',None,'Rudransh','Ayansh','Pratyush','Snehal'],
    'age':[23,None,24,26,25,29],
    'salary':[20000,None,45000,60000,34000,69000],
    'performance_score':[10,None,30,60,20,40]
}
df =pd.DataFrame(data)
df.fillna(0,inplace=True)
print(df)
```

```python
import pandas as pd 
data = {
    'name':['Arnav',None,'Rudransh','Ayansh','Pratyush','Snehal'],
    'age':[23,None,24,26,25,29],
    'salary':[20000,None,45000,60000,34000,69000],
    'performance_score':[10,None,30,60,20,40]
}
df =pd.DataFrame(data)
df.fillna({'age':df['age'].mean()},inplace=True)
print(df)
```

```python
df.fillna({'name':'Shekhar'},inplace=True)
df.fillna({'salary':df['salary'].mean()},inplace=True)
df.fillna({'performance_score':df['performance_score'].mean()},inplace=True)
print(df)
```
