```python
# using insert() 
import pandas as pd 
data = {
    'name':['Arnav','Rudransh','Ayansh','Pratyush','Snehal'],
    'age':[23,24,26,25,29],
    'salary':[20000,45000,60000,34000,69000],
    'performance_score':[10,30,60,20,40]
}
df =pd.DataFrame(data)

df.insert(1,'Employee_Id',[101,102,103,104,105])
print(df)
df_renamed = df.rename(
    columns={
        'name':'Employee_Name'
    }
)

print(df_renamed)
```
