```python
import numpy as np
arr = np.array([1,2,3,4,5])
print(arr[1])
```

```python
print(arr[4])
```

```python
# Slicing
print(arr[1:4:2])
```

```python
print(arr[1:4])
print(arr[:5])
print(arr[::2])
print(arr[::-1])
```

```python
# Fancy indexing
print(arr[[0,2,4]])
```

```python
# Filtering indexing / boolean masking
import numpy as np
arr = np.array([1,20,33,42,5])
print(arr[arr>25])
```
